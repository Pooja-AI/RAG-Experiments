| Category | Example libraries | Description |
|---|---|---|
| **RAG and agentic RAG orchestration** | LangChain, LangGraph, LlamaIndex | Orchestration frameworks bundle common RAG components like retrievers, generators, prompts, and memory into unified abstractions. They handle the plumbing for vector stores, LLMs, and databases, letting you focus on application logic rather than integration code. |
| **LLM/embeddings** | OpenAI, Anthropic, Transformers, Sentence Transformers | Foundation models and embedding services form the core of your RAG system, providing the intelligence for both understanding queries and generating responses. |
| **Vector storage** | Chroma, FAISS, Pinecone, Milvus, Weaviate | Vector databases enable efficient similarity search by storing and retrieving embeddings. |
| **Data processing** | pandas, NumPy, PyPDF2, pypdf, python-docx, Unstructured, openpyxl, scikit-learn | Libraries for data manipulation, file reading, and preprocessing tasks. Help load and preprocess unstructured data before connecting it to your RAG app. |
| **Multimodal and media** | Pillow, Pytesseract, MoviePy, pdf2image, opencv-python | Mature libraries for loading and processing various file formats including videos, podcasts, Word docs, PowerPoint presentations, and images. |
| **Text processing and NLP** | NLTK, Transformers, Rank-BM25, Beautiful Soup 4 | Traditional NLP libraries that help process text efficiently and perform text analysis tasks without relying solely on LLMs. |
| **Evaluation and monitoring** | Ragas, Phoenix, LangSmith, Prometheus-Eval | Frameworks providing predefined metrics to evaluate RAG app components (retriever, generator) and overall application performance. |
| **Web frameworks and deployment** | Streamlit, Gradio, Flask, Django | Frameworks for building user interfaces and deploying RAG applications. Streamlit and Gradio are ideal for rapid prototyping, while Flask and Django work well for production-grade apps. |
| **Database and storage** | SQLAlchemy, Psycopg2, SQLite3 | Connectors and object-relational mapping libraries for working with traditional SQL databases as data sources behind RAG apps. |
