#### Storing Secrets and API Keys with .env Files

Use an .env file to store secrets. This keeps API keys out of your codebase and prevents accidental commits to version control.

Add the .env file to your .gitignore. The .gitignore file lists files that Git should not track, ensuring that secrets stay local and never reach remote repositories.

```python
from dotenv import load_dotenv
import os
load_dotenv()
openai_key = os.getenv("OPENAI_API_KEY")
```

- Environment variables are key-value pairs that your program reads from the operating system at runtime. They provide a standard interface for accessing configuration and secrets without hardcoding values in source code.

- The .env approach works well for local development because it keeps secrets out of version control without requiring cloud infrastructure. However, .env files are plain text stored on disk, providing no encryption or access control.

- For production systems, use managed secret stores like Amazon Web Services (AWS) Secrets Manager, Microsoft Azure Key Vault, or HashiCorp Vault. These services provide encryption at rest and in transit, audit logging, automatic rotation, and fine-grained access controls. Most cloud platforms inject secrets directly as environment variables at runtime, so your application code remains unchanged—it still uses os.getenv.

- Don’t share .env files via email, Slack, or shared drives when multiple team members need access to the same secrets—this defeats the security benefits. Instead, use a shared development secret store or credential management tool like 1Password or AWS Secrets Manager with developer access.

- The key principle is that secrets should never be hardcoded in source code or committed to version control. Whether you use .env files locally or a managed service in production, environment variables provide a consistent interface for accessing secrets.
