###### Azure hosts models such as GPT-5, GPT-5 mini, GPT-5 nano, and text-embedding-3-small within the Azure ecosystem.

Each message consists of two required fields. The role field indicates who is speaking (for example, system, user, or assistant). The content field contains the actual message text.

```python
{
  "role": "user",
  "content": "What is RAG?"
}
```

```python
from openai import OpenAI

def ask_with_context(context, question):
    client = OpenAI()

    messages = [
        {
            "role": "system",
            "content": "Answer based only on the provided context."
        },
        {
            "role": "user",
            "content": f"Context:\n{context}\n\nQuestion:\n{question}"
        },
    ]

    response = client.chat.completions.create(
        model="gpt-5.2",
        messages=messages
    )

    return response.choices[0].message.content
```

```python
# Usage
context = "RAG stands for Retrieval-Augmented Generation."
question = "What does RAG stand for?"
answer = ask_with_context(context, question)
print(answer)
```
